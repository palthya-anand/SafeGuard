# 📊 Notebook 01 – Exploratory Data Analysis
This notebook explores the cleaned SafeGuard accident dataset to understand temporal patterns, weather effects, severity distribution, and geographic spread.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import folium
from folium.plugins import HeatMap
from pathlib import Path

sns.set_theme(style='whitegrid', palette='muted')
%matplotlib inline

DATA_DIR = Path('data/processed')
df = pd.read_csv(DATA_DIR / 'accidents_clean.csv')
print(f'Shape: {df.shape}')
df.head()
```

## 1. Accidents by Hour of Day
Peak accident hours reveal when roads are most dangerous. We expect spikes during morning and evening rush hours.

```python
fig, ax = plt.subplots(figsize=(12, 5))
hour_counts = df['hour'].value_counts().sort_index()
ax.bar(hour_counts.index, hour_counts.values, color='steelblue', edgecolor='white')
ax.set_xlabel('Hour of Day')
ax.set_ylabel('Number of Accidents')
ax.set_title('Accidents by Hour of Day')
ax.set_xticks(range(24))
plt.tight_layout()
plt.show()
```

## 2. Accidents by Day of Week
0 = Monday … 6 = Sunday. Weekends may show different patterns due to leisure travel and reduced traffic management.

```python
day_labels = ['Mon','Tue','Wed','Thu','Fri','Sat','Sun']
fig, ax = plt.subplots(figsize=(9, 5))
day_counts = df['day_of_week'].value_counts().sort_index()
ax.bar(day_labels, day_counts.values, color='coral', edgecolor='white')
ax.set_xlabel('Day of Week')
ax.set_ylabel('Number of Accidents')
ax.set_title('Accidents by Day of Week')
plt.tight_layout()
plt.show()
```

## 3. Accidents by Weather Condition
Adverse weather (rain, fog, storm) significantly increases accident probability.

```python
fig, ax = plt.subplots(figsize=(10, 5))
weather_counts = df['weather'].value_counts()
ax.bar(weather_counts.index, weather_counts.values, color='teal', edgecolor='white')
ax.set_xlabel('Weather Condition')
ax.set_ylabel('Number of Accidents')
ax.set_title('Accidents by Weather Condition')
plt.tight_layout()
plt.show()
```

## 4. Severity Distribution
Understanding the proportion of fatal vs. minor accidents helps prioritise intervention resources.

```python
fig, ax = plt.subplots(figsize=(7, 7))
sev_counts = df['severity'].value_counts()
colors = ['#2ecc71', '#f39c12', '#e74c3c', '#8e44ad']
ax.pie(
    sev_counts.values,
    labels=sev_counts.index,
    autopct='%1.1f%%',
    colors=colors[:len(sev_counts)],
    startangle=140,
)
ax.set_title('Accident Severity Distribution')
plt.tight_layout()
plt.show()
```

## 5. Road Type Breakdown
Highways and expressways tend to produce more severe accidents despite often being better maintained.

```python
fig, ax = plt.subplots(figsize=(9, 5))
road_counts = df['road_type'].value_counts()
ax.bar(road_counts.index, road_counts.values, color='mediumpurple', edgecolor='white')
ax.set_xlabel('Road Type')
ax.set_ylabel('Number of Accidents')
ax.set_title('Accidents by Road Type')
plt.tight_layout()
plt.show()
```

## 6. Monthly Accident Trend
A time-series view of accident counts across the dataset's date range.

```python
df['date'] = pd.to_datetime(df['date'])
monthly = df.groupby(df['date'].dt.to_period('M')).size()
fig, ax = plt.subplots(figsize=(14, 5))
monthly.plot(ax=ax, color='royalblue', marker='o', linewidth=1.5)
ax.set_xlabel('Month')
ax.set_ylabel('Number of Accidents')
ax.set_title('Monthly Accident Trend')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 7. Geographic Heatmap
Interactive Folium heatmap showing accident density across India.

```python
m = folium.Map(location=[20.5, 78.9], zoom_start=5, tiles='CartoDB positron')
heat_data = df[['latitude', 'longitude']].dropna().values.tolist()
HeatMap(heat_data, radius=8, blur=10).add_to(m)
m
```
